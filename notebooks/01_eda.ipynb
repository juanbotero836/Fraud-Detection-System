```python
# Import libraries

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('../data/raw/creditcard.csv')
print(df.shape)
df.head()
```

```python
df.info()
df.describe()
```

```python
balance = df['Class'].value_counts()
print(balance)
print(f'\nFraudes: {balance[1] / len(df) * 100:.4f} %')

balance.plot(kind='bar', color=['steelblue', 'crimson'])
plt.title('Distibucion de clases')
plt.xticks([0,1], ['Legitima', 'Fraude'], rotation=0)
plt.tight_layout()
plt.savefig('../notebooks/class_distribution.png')
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12,6))

df[df['Class'] == 0]['Amount'].hist(ax=axes[0], bins=50, color='steelblue')
axes[0].set_title('Monto - Transacciones Legitimas')
axes[0].set_xlabel('Monto')

df[df['Class'] == 1]['Amount'].hist(ax=axes[1], bins=50, color='crimson')
axes[1].set_title('Monto - Transacciones Fraudulentas')
axes[1].set_xlabel('Monto')

plt.tight_layout()
plt.savefig('../notebooks/amont_distribution.png')
plt.show()
```

```python
correlations = df.corr()['Class'].drop('Class').sort_values()

plt.figsize=(12,8)
correlations.plot(kind='barh', color=['crimson' if x < 0 else 'steelblue' for x in correlations])
plt.title('Correlacion de features con la clase (Fraude)')
plt.tight_layout
plt.savefig('../notebooks/feature_correlation.png')
plt.show()
```
